# Game IV: Loop Island - 民国教育委员会 （Question - Answer）


````python
# List of (question, answer) pairs
quiz = [
    ("What is the capital of France?", "Paris"),
    ("How many planets are in our solar system?", "8"),
    ("What is 5 + 7?", "12"),
    ("What color do you get by mixing blue and yellow?", "green"),
    ("What is the largest planet in our solar system?", "Jupiter"),
    ("How many sides does a hexagon have?", "6"),
    ("What is the chemical symbol for water?", "H2O"),
    ("Who wrote 'Romeo and Juliet'?", "Shakespeare"),
    ("What is the square root of 64?", "8"),
    ("Which Python function is used to get input from the user?", "input"),
]
num_of_questions = 5
quiz_selected = random.sample(quiz, num_of_questions)

for (question, correct_answer) in quiz_selected:
    print(f"{question}   {correct_answer}")
````

## System will select 5 question/answer from the database (code shown above), You will design the game
- Show question one by one to player, and get player's answer
- If player's answer is correct, he/she will get 1 score, otherwise, score 0, print a message to tell player
- At the end of the game, calculate the final total score of the player
- If the score is 5: "Perfect!"
- If the score is lager than 3: "Good job!"
