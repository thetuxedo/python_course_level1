## Section 1: String Basics — Decoding the Data Stream

---

### 🧠 What is a String?

A **string** is a sequence of characters — letters, numbers, symbols, spaces. In Python, strings are enclosed in quotes (single `'` or double `"`).

Think of a string like a **data stream** flowing through the neon-lit networks of Neo-Shanghai. Each character is a piece of information.

```python
# Creating strings
hacker_alias = "ZeroCool"
secret_msg = 'Meet at sector 7'
multiline = """This is a
multi-line transmission"""
```

### 🔢 String Indexing

Every character in a string has a **position (index)**, starting at `0`.

```python
code = "HACK3R"
#        H  A  C  K  3  R
#        0  1  2  3  4  5   (positive index)
#       -6 -5 -4 -3 -2 -1   (negative index)

print(code[0])    # H
print(code[4])    # 3
print(code[-1])   # R (last character)
```

### ✂️ String Slicing

You can extract a **substring** using slicing: `string[start:stop:step]`

```python
transmission = "OMNICORP_SECRET_FILE"

print(transmission[0:8])     # OMNICORP
print(transmission[9:15])    # SECRET
print(transmission[:4])      # OMNI (from beginning)
print(transmission[-4:])     # FILE (to end)
print(transmission[::2])     # Every 2nd char: OIOPSCEFILE
print(transmission[::-1])    # Reversed: ELIF_TERCES_PROCINMO
```

### 📏 String Length & Concatenation

```python
alias = "Ghost"
print(len(alias))           # 5

full_name = alias + "_" + "Runner"   # Concatenation
print(full_name)            # Ghost_Runner

echo = "beep " * 3          # Repetition
print(echo)                 # beep beep beep
```

### 🔄 Iterating Over Strings

```python
password = "N30N"
for char in password:
    print(char)
# N
# 3
# 0
# N
```

## Section 2: String Methods — Your Hacker Toolkit

---

### 🛠️ Essential String Methods

Python strings come with powerful built-in methods. Think of these as your hacking tools:

| Method | Description | Example |
|--------|-------------|--------|
| `.upper()` | ALL CAPS | `"hack".upper()` → `"HACK"` |
| `.lower()` | all lowercase | `"HACK".lower()` → `"hack"` |
| `.strip()` | Remove whitespace from edges | `" data ".strip()` → `"data"` |
| `.replace(old, new)` | Replace substrings | `"cat".replace("c","b")` → `"bat"` |
| `.split(sep)` | Split into list | `"a-b-c".split("-")` → `["a","b","c"]` |
| `.join(list)` | Join list into string | `"-".join(["a","b"])` → `"a-b"` |
| `.find(sub)` | Find index of substring (-1 if not found) | `"hello".find("ll")` → `2` |
| `.count(sub)` | Count occurrences | `"banana".count("a")` → `3` |
| `.startswith(s)` | Check prefix | `"Neo".startswith("Ne")` → `True` |
| `.endswith(s)` | Check suffix | `"file.exe".endswith(".exe")` → `True` |

### 💡 Important: Strings are IMMUTABLE!

You **cannot** change a string in place. Methods return a **new** string.

```python
msg = "hello"
msg.upper()       # Returns "HELLO" but doesn't change msg!
msg = msg.upper() # You must reassign to save the change

# This will cause an error:
# msg[0] = "H"   ← TypeError! Can't assign to string index
```

## Section 3: String Formatting — Crafting Digital Messages

---

### 📡 f-strings (Formatted String Literals)

The most modern and readable way to embed variables in strings. Use `f"...{variable}..."`

```python
agent = "ZeroCool"
sector = 7
status = "Active"

# f-string formatting
report = f"Agent {agent} is in Sector {sector}. Status: {status}."
print(report)
# Output: Agent ZeroCool is in Sector 7. Status: Active.
```

### 🧮 Expressions Inside f-strings

You can put any Python expression inside the curly braces!

```python
credits = 15000
cost = 4500

print(f"Credits remaining: {credits - cost}")
# Output: Credits remaining: 10500

name = "ghost"
print(f"Welcome, {name.upper()}!")
# Output: Welcome, GHOST!
```

### 📐 Formatting Numbers

```python
# Decimal places
hack_progress = 67.3333333
print(f"Progress: {hack_progress:.1f}%")
# Output: Progress: 67.3%

# Padding/alignment
for i in range(1, 4):
    print(f"Sector {i:02d}: Scanning...")
# Sector 01: Scanning...
# Sector 02: Scanning...
# Sector 03: Scanning...
```

### 📏 Padding & Alignment

```python
# Right-align with > (default for numbers)
agent = "Neo"
print(f"|{agent:>10}|")
# Output: |       Neo|

# Left-align with < (default for strings)
print(f"|{agent:<10}|")
# Output: |Neo       |

# Center-align with ^
print(f"|{agent:^10}|")
# Output: |   Neo    |

# Fill character + alignment
print(f"|{agent:*>10}|")
# Output: |*******Neo|

print(f"|{agent:=^10}|")
# Output: |===Neo====|

# Practical example: formatted table
agents = ["ZeroCool", "GhostFox", "NeonBlade"]
scores = [97, 82, 100]

print(f"{'AGENT':<12}{'SCORE':>6}")
print("-" * 18)
for i in range(len(agents)):
    print(f"{agents[i]:<12}{scores[i]:>6}")
# AGENT        SCORE
# ------------------
# ZeroCool         97
# GhostFox         82
# NeonBlade       100
```

template = "Target: {}. Threat Level: {}."

## Section 4: Advanced Strings — Parsing & Validation

---

### 🔍 Parsing Structured Strings

Real programs often receive several pieces of data packed into one string. You can split, inspect, validate, and rebuild that data with string methods.

```python
record = "agent=ZeroCool|sector=7|status=ACTIVE"
fields = record.split("|")
print(fields)
# ['agent=ZeroCool', 'sector=7', 'status=ACTIVE']
```

### ✂️ Split Only Where You Need To

```python
packet = "COMMAND:UPLOAD:secret.txt"
parts = packet.split(":", 1)
command = parts[0]
payload = parts[1]

print(command)  # COMMAND
print(payload)  # UPLOAD:secret.txt
```

### ✅ Validate String Content

```python
pin = "4821"
alias = "Ghost_7"

print(pin.isdigit())                         # True
print(alias.replace("_", "").isalnum())  # True
```

Useful checks include:

- `.isalpha()` — letters only
- `.isdigit()` — digits only
- `.isalnum()` — letters and digits only
- `.isspace()` — whitespace only
- `.startswith()` and `.endswith()` — required prefix or suffix

### 🔎 Find, Slice, and Rebuild

```python
email = "agent@glitch.net"
at_index = email.find("@")
name = email[:at_index]
domain = email[at_index + 1:]
print(name)       # agent
print(domain)     # glitch.net

safe_name = name.upper().replace(" ", "_")
rebuilt = safe_name + "@" + domain.lower()
print(rebuilt)    # AGENT@glitch.net
```
