### Problem Description

**BMI Formula:**

$$BMI = \frac{weight}{height^2}$$

where:
- `weight` is in kilograms (kg)
- `height` is in meters (m)

### Input
- The first variable contains the weight in kilograms (a real number)
- The second variable contains the height in meters (a real number)

### Output

- If BMI > 25.0, print: `Overweight`
- If 18.5 ≤ BMI ≤ 25.0, print: `Normal weight`
- If BMI < 18.5, print: `Underweight`

### Sample Input
```python
weigth = 69.0
height = 1.73
```

### Sample Output
```

```

### Explanation
