## Section 1: Why Do We Need Loops? (15 min)

### The Problem Without Loops

Imagine you want to print a countdown from 5 to 1 for your rocket launch. Without loops, you'd write:

````python
# Without loops - repetitive and boring!
print("5...")
print("4...")
print("3...")
print("2...")
print("1...")
print("🚀 LIFTOFF!")
````

That's 6 lines for just counting down from 5! What if you needed to count down from 100? That would be 101 lines! 😱

### The Solution: Loops!

A **loop** lets you repeat a block of code multiple times. Python has two main types:

| Loop Type | When to Use |
|-----------|-------------|
| `while` loop | When you don't know exactly how many times to repeat |
| `for` loop | When you know how many times to repeat, or want to go through a collection |

Let's see the same countdown with a loop:

````python
# With a loop - clean and powerful!
countdown = 5
while countdown > 0:
    print(f"{countdown}...")
    countdown = countdown - 1
print("🚀 LIFTOFF!")
````

## Section 2: While Loops (20 min)

### How While Loops Work

A `while` loop keeps repeating **as long as a condition is True**.

```python
while condition:
    # code to repeat
    # (don't forget to update something so the condition eventually becomes False!)
```

**Think of it like this:** "While there's still fuel in the tank, keep flying!"

### Key Parts of a While Loop:
1. **Initialization** - Set up your starting value
2. **Condition** - The test that's checked before each loop
3. **Body** - The code that repeats
4. **Update** - Change something so the loop eventually stops

⚠️ **WARNING:** If you forget the update step, you'll create an **infinite loop** and your program will run forever!

````python
# Example: Spaceship fuel consumption
fuel = 100  # 1. Initialization

while fuel > 0:  # 2. Condition
    print(f"Fuel level: {fuel}% - Flying through space... 🚀")  # 3. Body
    fuel = fuel - 25  # 4. Update

print("⚠️ Out of fuel! Time to find a space station!")
````

````python
# Example: Collecting space crystals until cargo is full
cargo = 0
cargo_max = 5

print("Starting crystal collection mission! 💎")
while cargo < cargo_max:
    cargo = cargo + 1
    print(f"  Crystal collected! Cargo: {cargo}/{cargo_max}")

print("Cargo full! Heading back to base. 🏠")
````

### Using While Loops with User Input

While loops are great for repeatedly asking the user for input:

````python
# Example: Space station access code
password = ""
secret_code = "GALAXY"

while password != secret_code:
    password = input("Enter the space station access code: ")
    if password != secret_code:
        print("❌ Access denied! Try again.")

print("✅ Access granted! Welcome aboard the space station!")
````

## Section 3: For Loops and Range (25 min)

### The `for` Loop

A `for` loop goes through each item in a **sequence** (like a list, string, or range of numbers).

```python
for item in sequence:
    # do something with item
```

**Think of it like this:** "For each planet in my star map, scan it!"

````python
# Example: Scanning planets
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter"]

for planet in planets:
    print(f"🔭 Scanning {planet}...")

print("\nAll planets scanned! ✅")
````

````python
# Example: Looping through a string (each character)
ship_name = "NOVA"

print("Ship ID code:")
for letter in ship_name:
    print(f"  [{letter}]")
````

### The `range()` Function

The `range()` function generates a sequence of numbers. It's your best friend with `for` loops!

| Syntax | What it does | Example |
|--------|-------------|----------|
| `range(stop)` | Numbers from 0 to stop-1 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | Numbers from start to stop-1 | `range(2, 6)` → 2,3,4,5 |
| `range(start, stop, step)` | Numbers from start to stop-1, skipping by step | `range(0, 10, 2)` → 0,2,4,6,8 |

````python
# Example: range(stop)
print("Activating shields:")
for i in range(5):
    print(f"  Shield layer {i} activated!")
````

````python
# Example: range(start, stop)
print("Decks 3 through 7 status check:")
for deck in range(3, 8):
    print(f"  Deck {deck}: All clear! ✅")
````

````python
# Example: range(start, stop, step)
print("Power levels (increasing by 20):")
for power in range(0, 101, 20):
    print(f"  Power: {power}%")
````

````python
# Example: Counting backwards with negative step
print("Countdown using for loop:")
for i in range(5, 0, -1):
    print(f"  {i}...")
print("  Blast off! 🚀")
````

### Accumulator Pattern

A common pattern is using a loop to **accumulate** (build up) a value:

````python
# Example: Calculating total distance traveled
daily_distances = [150, 200, 175, 220, 190]  # light-years per day
total_distance = 0

for distance in daily_distances:
    total_distance = total_distance + distance
    print(f"  Traveled {distance} ly today. Total: {total_distance} ly")

print(f"\n📍 Total mission distance: {total_distance} light-years!")
````

## Section 4: Break and Continue (20 min)

Sometimes you need more control over your loops:

- **`break`** - Immediately EXIT the loop (escape!)
- **`continue`** - SKIP the rest of this iteration and go to the next one

### The `break` Statement

Use `break` when you want to stop the loop early based on some condition.

````python
# Example: Searching for a habitable planet
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn"]

print("Searching for habitable planet...")
for planet in planets:
    print(f"  Checking {planet}...")
    if planet == "Earth":
        print(f"  ✅ Found it! {planet} is habitable!")
        break  # Stop searching, we found it!

print("Search complete.")
````

````python
# Example: Emergency stop - asteroid detected!
print("Traveling through asteroid field...")
for sector in range(1, 11):
    if sector == 6:
        print(f"  ☄️ DANGER! Giant asteroid in sector {sector}! Emergency stop!")
        break
    print(f"  Sector {sector}: Clear ✅")

print("Navigation paused.")
````

### The `continue` Statement

Use `continue` when you want to skip certain items but keep looping.

````python
# Example: Skip damaged sectors during scan
damaged_sectors = [3, 5, 7]

print("Scanning all 8 sectors (skipping damaged ones):")
for sector in range(1, 9):
    if sector in damaged_sectors:
        print(f"  Sector {sector}: ⚠️ DAMAGED - Skipping!")
        continue  # Skip to next sector
    print(f"  Sector {sector}: Scanned successfully ✅")
````

````python
# Example: Only process even-numbered transmissions
print("Processing transmissions (only even-numbered):")
for transmission in range(1, 11):
    if transmission % 2 != 0:  # If odd number
        continue  # Skip odd transmissions
    print(f"  📡 Processing transmission #{transmission}")
````

## Section 5: Nested Loops (20 min)

### Loops Inside Loops!

You can put a loop inside another loop. The **inner loop** runs completely for each iteration of the **outer loop**.

**Think of it like this:** For each deck on the spaceship, check every room on that deck.

````python
# Example: Checking rooms on each deck
print("Ship inspection report:")
for deck in range(1, 4):  # 3 decks
    print(f"\n📋 Deck {deck}:")
    for room in range(1, 4):  # 3 rooms per deck
        print(f"    Room {room}: Inspected ✅")
````

````python
# Example: Multiplication table for navigation calculations
print("Speed x Time = Distance Table")
print("=" * 30)
for speed in range(1, 4):
    for time in range(1, 6):
        distance = speed * time
        print(f"  {speed} x {time} = {distance}")
    print()  # Empty line between groups
````

````python
# Example: Grid map of space
print("Space Grid Map (5x5):")
for row in range(5):
    for col in range(5):
        if row == 2 and col == 2:
            print("🚀", end=" ")  # Our ship in the center!
        else:
            print("· ", end=" ")
    print()  # New line after each row
````
