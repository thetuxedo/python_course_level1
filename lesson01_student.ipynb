## Section 2: Variables — Storage Boxes with Name Tags 📦
*Estimated time: 20 minutes*

---

### What is a Variable?

Think of a variable like a **storage box with a name tag**:
- The **name tag** is the variable name
- The **stuff inside** is the value
- You can **change** what's inside anytime!

### The `=` sign means "put this value into the box"

```python
variable_name = value
```

### Types of Data You Can Store:

| Type | Example | Description |
|------|---------|-------------|
| Integer (int) | `my_age = 12` | Whole numbers |
| Float | `price = 9.99` | Decimal numbers |
| String (str) | `name = "Loki"` | Text (always in quotes) |
| Boolean (bool) | `is_hungry = True` | True or False |

````python
# Creating variables - like labeling storage boxes!
my_age = 12
pet_name = "Loki"
favorite_number = 8
is_hungry = True

# Let's see what's inside our boxes
print(my_age)
print(pet_name)
print(favorite_number)
print(is_hungry)
````

### Combining Text and Variables in `print()`

There are two ways to print variables with text:

**Method 1: Using `+` (concatenation)** — joins strings together
```python
print("Hello, " + pet_name)
```

**Method 2: Using commas** — separates items with a space
```python
print("Hello,", pet_name)
```

````python
pet_name = "Loki"

# Method 1: Using + (only works with strings)
print("Hello, " + pet_name)

# Method 2: Using commas (works with any type)
print("Hello,", pet_name)
print("My age is", my_age)
````

### Changing Variables

You can change what's inside a variable at any time — just assign a new value!

````python
# Variables can change!
pet_name = "Loki"
print("My pet is", pet_name)

pet_name = "Fluffy"  # Changed the value!
print("Now my pet is", pet_name)
````

````python
# A common pattern: updating a variable based on its current value
score = 0
print("Starting score:", score)

score = score + 10  # Add 10 to the current score
print("After bonus:", score)

score = score + 5   # Add 5 more
print("After another bonus:", score)
````

### ⚠️ Variable Naming Rules:
- ✅ Can contain letters, numbers, and underscores: `my_score2`
- ✅ Must start with a letter or underscore: `_name`, `age`
- ❌ Cannot start with a number: `2name` is INVALID
- ❌ Cannot contain spaces: `my name` is INVALID (use `my_name`)
- ❌ Cannot use Python keywords: `print`, `if`, `for` are reserved
- ⚠️ Python is case-sensitive: `Name` and `name` are DIFFERENT variables

## Section 3: Python the Super Calculator 🧮
*Estimated time: 20 minutes*

---

Python can do math! Here are the arithmetic operators:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `10 - 4` | `6` |
| `*` | Multiplication | `3 * 7` | `21` |
| `/` | Division | `15 / 4` | `3.75` |
| `//` | Floor Division (whole part) | `15 // 4` | `3` |
| `%` | Modulo (remainder) | `15 % 4` | `3` |
| `**` | Exponent (power) | `2 ** 3` | `8` |

````python
# Basic arithmetic
print("Addition: 5 + 3 =", 5 + 3)
print("Subtraction: 10 - 4 =", 10 - 4)
print("Multiplication: 3 * 7 =", 3 * 7)
print("Division: 15 / 4 =", 15 / 4)
````

````python
# Special operators
print("Floor Division: 15 // 4 =", 15 // 4)  # Removes the decimal part
print("Remainder: 15 % 4 =", 15 % 4)         # What's left over
print("Power: 2 ** 3 =", 2 ** 3)              # 2 × 2 × 2
````

### 🍪 Real-World Example: Sharing Cookies

### 🛒 Real-World Example: Shopping

````python
price = 15
quantity = 3
total = price * quantity
print("Total cost: $" + str(total))
````

### 🔑 Understanding `%` (Modulo) and `//` (Floor Division)

These are super useful! Think of dividing 13 by 4:
- `13 / 4 = 3.25` (exact answer)
- `13 // 4 = 3` (how many whole times does 4 go into 13?)
- `13 % 4 = 1` (what's left over after dividing?)

**Real-world example:** You have 13 candies and 4 bags:
- Each bag gets `13 // 4 = 3` candies
- Leftover candies: `13 % 4 = 1`

````python
print("13 / 4 =", 13 / 4)    # Exact division
print("13 // 4 =", 13 // 4)  # Floor division (whole part)
print("13 % 4 =", 13 % 4)    # Remainder

print()
print("9 ** 2 =", 9 ** 2)    # 9 squared = 81
````

### 💡 Tip: Checking Even or Odd

A number is **even** if `number % 2 == 0` (no remainder when divided by 2)

A number is **odd** if `number % 2 == 1`

````python
print("10 % 2 =", 10 % 2)  # 0 → even!
print("7 % 2 =", 7 % 2)    # 1 → odd!
````

### The `input()` Function — Let's Have a Conversation!

The `input()` function:
1. Displays a message (prompt) to the user
2. **Waits** for the user to type something
3. Returns what they typed as a **string**

```python
variable = input("Your prompt message: ")
```

### ⚠️ Important: `input()` ALWAYS returns a string!
Even if the user types a number, it comes back as text. To use it as a number, you must convert it:
- `int()` — converts to a whole number
- `float()` — converts to a decimal number

````python
# Basic input - getting a name
name = input("What is your name? ")
print("Hello, " + name + "!")
````

````python
# Getting a number from the user
# We use int() to convert the text input to a number
age = int(input("How old are you? "))
print("Wow,", age, "is a great age!")
print("In 5 years, you'll be", age + 5)
````

### Why do we need `int()`?

Without converting, you can't do math with input!

````python
# This shows the difference:
num_text = "5"         # This is a STRING "5"
num_real = int("5")    # This is the NUMBER 5

print(num_text + num_text)  # String + String = "55" (joins them)
print(num_real + num_real)  # Number + Number = 10 (adds them)
````

## Section 5: If-Else Statements — Making Decisions 🚦
*Estimated time: 20 minutes*

---

### Programs Can Make Decisions!

Every day YOU make decisions:
- **IF** it's raining → bring an umbrella
- **ELSE** → wear sunglasses

Python can do the same thing!

### The Structure:

```python
# Simple if:
if CONDITION:
    do something

# If-else:
if CONDITION:
    do something
else:
    do something different

# If-elif-else (multiple conditions):
if CONDITION_1:
    do thing 1
elif CONDITION_2:
    do thing 2
elif CONDITION_3:
    do thing 3
else:
    do default thing
```

### Comparison Operators:

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `x == 5` |
| `!=` | Not equal to | `x != 5` |
| `>` | Greater than | `x > 5` |
| `<` | Less than | `x < 5` |
| `>=` | Greater than or equal | `x >= 5` |
| `<=` | Less than or equal | `x <= 5` |

### ⚠️ Important:
- `=` means **assign** (put value in variable)
- `==` means **compare** (check if equal)
- Don't forget the **colon** `:` after the condition!
- The code inside must be **indented** (4 spaces)

````python
# Simple if-else: Even or Odd?
num = int(input("Give me a number: "))

if num % 2 == 0:
    print(num, "is an EVEN number")
else:
    print(num, "is an ODD number")
````

````python
# Using elif for multiple conditions
temperature = int(input("What's the temperature? "))

if temperature >= 30:
    print("It's HOT! 🔥 Stay hydrated!")
elif temperature >= 20:
    print("It's warm and nice! ☀️")
elif temperature >= 10:
    print("It's cool. Wear a jacket! 🧥")
else:
    print("It's COLD! ❄️ Bundle up!")
````

### Combining Conditions with `and` / `or`

- `and` — **both** conditions must be True
- `or` — **at least one** condition must be True

````python
age = int(input("How old are you? "))
has_ticket = input("Do you have a ticket? (yes/no) ")

if age >= 12 and has_ticket == "yes":
    print("Welcome to the movie! 🎬")
elif age < 12:
    print("Sorry, you must be 12 or older.")
else:
    print("You need a ticket to enter!")
````

### New Concepts Used:

### 🎮 Game 1: Guess The Number

The computer picks a secret number between 1 and 100. You have 10 tries to guess it!

The computer will tell you if your guess is too high or too low.

````python
import random

secret_num = random.randint(1, 100)
print("🎲 I'm thinking of a number between 1 and 100!")
print("You have 10 tries to guess it.\n")

for i in range(10):
    guess = int(input("Your guess: "))
    
    if guess == secret_num:
        print("🎉 Bingo! You found the number in", i + 1, "tries!")
        break
    elif guess > secret_num:
        print("📉 Too HIGH! Try a lower number.")
    else:
        print("📈 Too LOW! Try a higher number.")
else:
    print("\n😔 Out of tries! The number was", secret_num)
````

### 🎮 Game 2: Rock, Paper, Scissors ✊📄✂️

Play the classic game against the computer!

````python
import random

choices = ["rock", "paper", "scissors"]
computer = random.choice(choices)

player = input("Choose rock/paper/scissors: ")
player = player.lower()  # Convert to lowercase

print("\n👤 You:", player)
print("🤖 Computer:", computer)
print()

if player == computer:
    print("🤝 It's a TIE!")
elif player == "rock" and computer == "scissors":
    print("🎉 You WIN! Rock smashes scissors!")
elif player == "paper" and computer == "rock":
    print("🎉 You WIN! Paper covers rock!")
elif player == "scissors" and computer == "paper":
    print("🎉 You WIN! Scissors cut paper!")
else:
    print("😔 Computer wins!")
````
