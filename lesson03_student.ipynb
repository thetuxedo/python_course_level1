## Knowledge review

A **variable** is a name that points to a value.

```python
ship_name = "Python Pearl"  # str: text
crew_size = 8               # int: whole number
speed = 12.5                # float: decimal number
has_map = True              # bool: True or False
```

Useful ideas:

- `=` assigns a value; it does not mean “is equal to.”
- `type(value)` tells us a value's data type.
- Arithmetic operators include `+`, `-`, `*`, `/`, `//`, `%`, and `**`.
- `+` adds numbers but joins strings.
- Clear variable names make code easier to read.

# 2. 🌩️ Storm Strait — `if`, `elif`, and `else`
**Suggested time: 24 minutes**

## Knowledge review

Conditionals let a program choose a path.

```python
if wind_speed >= 60:
    print("Stay in the harbor!")
elif wind_speed >= 30:
    print("Sail carefully.")
else:
    print("Full speed ahead!")
```

Remember:

- Conditions evaluate to `True` or `False`.
- Use `==` to compare and `=` to assign.
- Python checks branches from top to bottom and runs the **first** true branch.
- `and` requires both conditions; `or` requires at least one; `not` reverses a Boolean.
- Indented code belongs to its conditional branch.

# 3. 🪸 Coral Maze — `for` Loops and `continue`
**Suggested time: 24 minutes**

## Knowledge review

A `for` loop repeats once for every value in a sequence.

```python
for island_number in range(1, 6):
    print("Scanning island", island_number)
```

About `range(start, stop, step)`:

- `start` is included.
- `stop` is excluded.
- `step` tells Python how much to change each time.
- `range(1, 6)` produces 1, 2, 3, 4, 5.

Use an **accumulator** to keep a running total. Use `continue` to skip the rest of one iteration without ending the loop.

# 4. 🐙 Kraken Trench — `while` Loops and `break`
**Suggested time: 24 minutes**

## Knowledge review

A `while` loop repeats **while its condition is true**.

```python
distance = 30
while distance > 0:
    print(distance, "miles left")
    distance -= 10
```

Safety checklist:

1. Initialize the control variable before the loop.
2. Check a condition that can eventually become false.
3. Update the control variable inside the loop.

Use `break` to leave the nearest loop immediately. Use `continue` carefully in a `while` loop: update the control variable **before** continuing, or the loop might run forever.
