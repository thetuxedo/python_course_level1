# 🔐 MISSION 1 — Tuples: Locking Down Coordinates
## *"Wildhaven Island protects all supply-cache locations. Once a location is encoded, it cannot be changed."*

---

## 📡 What is a Tuple?

A **tuple** is like a list — but **immutable** (it cannot be changed after creation).

Think of a tuple as a **locked field record** in Wildhaven Island. Once you encode the supply-cache coordinates, they are frozen.

### ✅ Creating a Tuple

```python
# Using parentheses
cache_location = (12, 47, 'Wildhaven-East')

# Without parentheses (tuple packing)
cache_location = 12, 47, 'Wildhaven-East'

# Single-element tuple — the comma is REQUIRED!
lone_resource = ('River',)   # ✅ This is a tuple
not_a_tuple = ('River')  # ❌ This is just a string!

# Empty tuple
empty_sector = ()
```

### 🔍 Accessing Elements (same as lists!)

```python
cache_location = (12, 47, 'Wildhaven-East')

print(cache_location[0])   # 12
print(cache_location[-1])  # 'Wildhaven-East'
print(cache_location[1:])  # (47, 'Wildhaven-East')
```

### 🚫 Tuples are IMMUTABLE

```python
cache_location = (12, 47, 'Wildhaven-East')
cache_location[0] = 99  # ❌ TypeError: 'tuple' object does not support item assignment
```

### 📦 Tuple Methods

Tuples have only **two built-in methods**:

| Method | Description | Example |
|--------|-------------|---------|
| `.count(x)` | Counts how many times `x` appears | `(1,2,2,3).count(2)` → `2` |
| `.index(x)` | Returns the index of the first `x` | `(1,2,3).index(3)` → `2` |

### 📊 Tuple vs. List — Quick Comparison

| Feature | List `[]` | Tuple `()` |
|---------|-----------|------------|
| Mutable? | ✅ Yes | ❌ No |
| Ordered? | ✅ Yes | ✅ Yes |
| Duplicates? | ✅ Yes | ✅ Yes |
| Methods | Many | Only 2 |
| Speed | Slower | ⚡ Faster |
| Use case | Dynamic data | Fixed/constant data |

### 💡 When to use a Tuple?
- Data that **should NOT change** (GPS coordinates, RGB colors, fixed field records)
- Slightly **faster** than lists
- Can be used as **dictionary keys** (lists cannot!)

````python
# 🖥️ LIVE DEMO — Run this cell to see tuples in action!

# SCOUT's weatherproof supply-cache coordinates
cache = (12, 47, 'Wildhaven-East', 'Level-5')

print("=== SUPPLY-CACHE COORDINATES ===")
print("Full record:", cache)
print("X coord:", cache[0])
print("Y coord:", cache[1])
print("Zone:", cache[2])
print("Supply level:", cache[3])
print("Number of fields:", len(cache))

# Slicing
print("\nCoordinates only:", cache[:2])

# Iterating
print("\n=== SCANNING PACKET ===")
for field in cache:
    print("  Field:", field)

# Methods
signal_log = (5, 3, 5, 8, 5, 3, 1)
print("\n=== SIGNAL LOG ===")
print("Signal log:", signal_log)
print("How many times signal '5' appeared:", signal_log.count(5))
print("First occurrence of signal '8' is at index:", signal_log.index(8))
````

# 🔁 MISSION 2 — Tuple Operations & Unpacking
## *"Advanced survivors know how to unpack weatherproof field records in a single line."*

---

## 🔓 Tuple Unpacking

**Tuple unpacking** lets you assign each element of a tuple to a separate variable in one line. This is one of Python's most powerful features!

```python
# Without unpacking (old way)
record = ('SCOUT', 42, 'Wildhaven-West')
name = record[0]
level = record[1]
zone = record[2]

# WITH unpacking (survivor way! ✅)
name, level, zone = record
print(name)   # SCOUT
print(level)  # 42
print(zone)   # Wildhaven-West
```

### ⚡ Star Unpacking — Grab Multiple Items at Once

```python
resources = ('Alpha', 'Beta', 'Gamma', 'Delta', 'Omega')

first, *middle, last = resources
print(first)   # Alpha
print(middle)  # ['Beta', 'Gamma', 'Delta']
print(last)    # Omega
```

## ➕ Tuple Concatenation & Repetition

Even though tuples are immutable, you can **create new tuples** by combining them:

```python
north_caches = ('CACHE-01', 'CACHE-02')
south_caches = ('CACHE-03', 'CACHE-04')

# Concatenation with +
all_caches = north_caches + south_caches
print(all_caches)  # ('CACHE-01', 'CACHE-02', 'CACHE-03', 'CACHE-04')

# Repetition with *
alarm = ('ALERT!',) * 3
print(alarm)  # ('ALERT!', 'ALERT!', 'ALERT!')
```

## 🔄 Converting Between List and Tuple

```python
# List → Tuple (lock it down)
my_list = [1, 2, 3]
my_tuple = tuple(my_list)

# Tuple → List (unlock for editing)
my_tuple = (1, 2, 3)
my_list = list(my_tuple)
my_list.append(4)
my_tuple = tuple(my_list)  # lock again
print(my_tuple)  # (1, 2, 3, 4)
```

## 🧪 Useful: `enumerate()` with Tuples

```python
caches = ('Alpha', 'Beta', 'Gamma')
for index, name in enumerate(caches):
    print(f"[{index}] Resource: {name}")
# [0] Resource: Alpha
# [1] Resource: Beta
# [2] Resource: Gamma
```

````python
# 🖥️ LIVE DEMO — Tuple Unpacking & Operations

# --- Unpacking ---
survivor_profile = ('SCOUT', 99, 'Wildhaven-Central', True)
name, skill_level, zone, is_active = survivor_profile

print("=== SURVIVOR PROFILE UNPACKED ===")
print(f"  Name: {name}")
print(f"  Skill Level: {skill_level}")
print(f"  Zone: {zone}")
print(f"  Active: {is_active}")

# --- Star Unpacking ---
trek_path = ('Beach Camp', 'Palm Grove', 'River Ford', 'Bamboo Trail', 'Signal Peak')
start, *waypoints, destination = trek_path

print("\n=== INFILTRATION ROUTE ===")
print(f"  From: {start}")
print(f"  Via: {waypoints}")
print(f"  To: {destination}")

# --- Concatenation ---
team_alpha = ('River', 'Fern')
team_beta  = ('Ember', 'Stone')
full_team  = team_alpha + team_beta
print("\n=== FULL EXPEDITION TEAM ===")
print(full_team)

# --- Enumerate ---
print("\n=== TEAM ROSTER ===")
for i, survivor in enumerate(full_team):
    print(f"  [{i}] {survivor}")
````

# 🔥 MISSION 3 — Sets: The Unique Resources
## *"Wildhaven's field journal keeps a registry of unique active resources. Duplicates are auto-purged."*

---

## 📡 What is a Set?

A **set** is an **unordered** collection of **unique** elements. Duplicates are automatically removed!

Think of a set as **Wildhaven's resource registry** — each region resource can only appear **once** in the system.

### ✅ Creating a Set

```python
# Using curly braces
active_resources = {'Alpha', 'Beta', 'Gamma', 'Alpha'}  # 'Alpha' appears once!
print(active_resources)  # {'Alpha', 'Beta', 'Gamma'} — order may vary!

# Using set() constructor
active_resources = set(['Alpha', 'Beta', 'Gamma', 'Alpha'])

# Empty set — MUST use set(), NOT {}
empty = set()    # ✅ Empty set
not_empty = {}   # ❌ This creates an empty DICTIONARY!
```

### ⚠️ Key Properties of Sets
- **No duplicates** — automatically removed
- **Unordered** — no indexing (`my_set[0]` will cause an error!)
- **Mutable** — you can add/remove elements
- Elements must be **immutable** (strings, numbers, tuples — but NOT lists)

### 🔧 Set Methods

| Method | Description | Example |
|--------|-------------|---------|
| `.add(x)` | Adds element `x` | `s.add('Delta')` |
| `.remove(x)` | Removes `x` (error if not found) | `s.remove('Alpha')` |
| `.discard(x)` | Removes `x` (no error if missing) | `s.discard('River')` |
| `.pop()` | Removes and returns a random element | `s.pop()` |
| `.clear()` | Removes all elements | `s.clear()` |
| `.copy()` | Returns a copy of the set | `s2 = s.copy()` |
| `len(s)` | Number of elements | `len(s)` |
| `x in s` | Membership test | `'Alpha' in s` |

### 📊 Set vs. List vs. Tuple

| Feature | List `[]` | Tuple `()` | Set `{}` |
|---------|-----------|------------|----------|
| Ordered? | ✅ | ✅ | ❌ |
| Duplicates? | ✅ | ✅ | ❌ |
| Mutable? | ✅ | ❌ | ✅ |
| Indexing? | ✅ | ✅ | ❌ |
| Fast lookup? | ❌ Slow | ❌ Slow | ⚡ Fast |

````python
# 🖥️ LIVE DEMO — Sets in Action!

# --- Creating sets ---
# Wildhaven Island's raw scan picks up duplicate resource IDs
raw_scan = ['Resource-A', 'Resource-B', 'Resource-C', 'Resource-A', 'Resource-D', 'Resource-B']
print("Raw scan (with duplicates):", raw_scan)

# Convert to set — duplicates vanish!
unique_resources = set(raw_scan)
print("Unique resources (set):", unique_resources)
print("Unique count:", len(unique_resources))

# --- Add and remove ---
print("\n=== MANAGING THE LANDMARK REGISTRY ===")
registry = {'Resource-A', 'Resource-B', 'Resource-C'}
print("Initial registry:", registry)

registry.add('Resource-E')          # Add new resource
print("After add Resource-E:", registry)

registry.remove('Resource-B')       # Remove Resource-B
print("After remove Resource-B:", registry)

registry.discard('Resource-Z')      # Discard non-existent — no error!
print("After discard Resource-Z (safe):", registry)

# --- Membership testing ---
print("\n=== TEAM CHECK ===")
print("Is Resource-A in registry?", 'Resource-A' in registry)
print("Is Resource-B in registry?", 'Resource-B' in registry)

# --- Iterating ---
print("\n=== ACTIVE NODES ===")
for resource in registry:
    print("  >", resource)
````

# ⚔️ MISSION 4 — Set Operations: Mapping the Island
## *"The most powerful survival techniques involve comparing island maps — finding what's shared, what's different, what's unique."*

---

## 🔗 Set Operations — The Survivor's Toolkit

Sets support powerful **mathematical operations** that are incredibly useful for comparing data:

### 1. 🤝 Union — All resources from BOTH regions
```python
# Everything in A OR B (or both)
a = {'Alpha', 'Beta', 'Gamma'}
b = {'Gamma', 'Delta', 'Epsilon'}

print(a | b)           # Operator
print(a.union(b))      # Method
# {'Alpha', 'Beta', 'Gamma', 'Delta', 'Epsilon'}
```

### 2. 🔍 Intersection — Resources in BOTH regions
```python
# Everything in A AND B
print(a & b)                # Operator
print(a.intersection(b))    # Method
# {'Gamma'}
```

### 3. ➖ Difference — Resources ONLY in A (not in B)
```python
# Everything in A but NOT in B
print(a - b)             # Operator
print(a.difference(b))   # Method
# {'Alpha', 'Beta'}
```

### 4. ↔️ Symmetric Difference — Resources in ONE but NOT BOTH
```python
# Everything in A or B, but NOT in both
print(a ^ b)                          # Operator
print(a.symmetric_difference(b))      # Method
# {'Alpha', 'Beta', 'Delta', 'Epsilon'}
```

### 5. 🔎 Subset & Superset
```python
x = {'Alpha', 'Beta'}
y = {'Alpha', 'Beta', 'Gamma'}

print(x.issubset(y))    # True — all of x is in y
print(y.issuperset(x))  # True — y contains all of x
print(x.isdisjoint(y))  # False — they share elements
```

### 🗺️ Visual Summary
```
A = {1, 2, 3}      B = {3, 4, 5}

  A ∪ B  = {1, 2, 3, 4, 5}   ← Union (all)
  A ∩ B  = {3}               ← Intersection (shared)
  A − B  = {1, 2}            ← Difference (A only)
  A △ B  = {1, 2, 4, 5}      ← Symmetric Difference (not shared)
```

````python
# 🖥️ LIVE DEMO — Set Operations

# Wildhaven Island's two island regions have different active resources
region_north = {'Resource-A', 'Resource-B', 'Resource-C', 'Resource-D', 'Resource-E'}
region_south = {'Resource-C', 'Resource-D', 'Resource-F', 'Resource-G'}

print("North Region:", region_north)
print("South Region:", region_south)

print("\n=== SET OPERATIONS ===")

# Union — all resources across both DCs
all_resources = region_north | region_south
print("All resources (Union):", all_resources)

# Intersection — resources in BOTH DCs (shared infrastructure)
shared_resources = region_north & region_south
print("Shared resources (Intersection):", shared_resources)

# Difference — resources ONLY in North
north_only = region_north - region_south
print("North-only resources (Difference):", north_only)

# Symmetric Difference — resources in one DC only
exclusive_resources = region_north ^ region_south
print("Exclusive resources (Sym. Diff.):", exclusive_resources)

# Subset check
core_resources = {'Resource-C', 'Resource-D'}
print("\nAre core resources a subset of North Region?", core_resources.issubset(region_north))
print("Are core resources a subset of South Region?", core_resources.issubset(region_south))
````

## 📋 What You Learned Today

### 🔐 Tuples
| Concept | Key Point |
|---------|----------|
| Creation | `t = (1, 2, 3)` or `t = 1, 2, 3` |
| Single element | `t = (1,)` — comma required! |
| Immutable | Cannot change elements after creation |
| Indexing/Slicing | Same as lists: `t[0]`, `t[1:]` |
| Methods | Only `.count()` and `.index()` |
| Unpacking | `a, b, c = t` |
| Star unpacking | `first, *rest = t` |
| Concatenation | `t1 + t2` creates a new tuple |
| When to use | Fixed/constant data, faster than lists |

### 🔥 Sets
| Concept | Key Point |
|---------|----------|
| Creation | `s = {'a', 'b', 'c'}` |
| Empty set | `s = set()` — NOT `{}` |
| Unique | Duplicates auto-removed |
| Unordered | No indexing! |
| Add | `.add(x)` |
| Remove (safe) | `.discard(x)` |
| Remove (strict) | `.remove(x)` |
| Union | `a \| b` or `a.union(b)` |
| Intersection | `a & b` or `a.intersection(b)` |
| Difference | `a - b` or `a.difference(b)` |
| Sym. Difference | `a ^ b` or `a.symmetric_difference(b)` |
| When to use | Unique items, fast membership tests, set math |

---

## 🧩 The Big Picture
| Feature | `list` | `tuple` | `set` |
| :--- | :--- | :--- | :--- |
| **Syntax** | `[1, 2, 3]` | `(1, 2, 3)` | `{1, 2, 3}` or `set()` |
| **Ordered?** | ✅ Yes (keeps insertion order) | ✅ Yes (keeps insertion order) | ❌ No (unordered, random display) |
| **Mutable?** | ✅ Yes (edit anytime) | ❌ No (immutable after creation) | ✅ Yes (add/remove items) |
| **Duplicates?** | ✅ Allowed | ✅ Allowed | ❌ Auto-removed (unique only) |
| **Index/Slice?** | ✅ `lst[0]`, `lst[1:3]` | ✅ `tup[0]`, `tup[1:3]` | ❌ Not supported |
| **`in` Lookup Speed** | 🐢 `O(n)` (checks each item) | 🐢 `O(n)` | 🚀 `O(1)` (hash-based, instant) |
| **Hashable?** | ❌ No | ✅ Yes (if contents are hashable) | ❌ No |
| **Best For** | Collections that grow/shrink (scores, to-dos) | Fixed data, coordinates, safe configs, function returns | Unique items, removing duplicates, finding overlaps |
