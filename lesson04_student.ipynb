# 🏰 Section 1 — What is a List?
**⏱ ~15 minutes**

## 📖 Concept: Why Do We Need Lists?

Imagine your knight has collected 5 magical items during the quest. Without lists, you'd need:

```python
item1 = "Sword"
item2 = "Shield"
item3 = "Potion"
item4 = "Map"
item5 = "Torch"
```

That's messy! 😩 A **list** lets you store multiple values in **one variable**:

```python

```

### 🔑 Key Facts
- A list is written with **square brackets** `[ ]`
- Items are separated by **commas** `,`
- A list can hold **different types**: strings, integers, floats, even other lists!
- Lists are **ordered** — the order you put items in is preserved

```python
# A mixed list
hero = ["Aria", 15, 100.0, True]
#        name  age   HP   alive?
```

# 🗡️ Section 2 — Indexing & Slicing
**⏱ ~20 minutes**

## 📖 Concept: Accessing Items by Index

Every item in a list has an **index** (its position number). Python starts counting from **0**!

```
inventory = ["Sword",  "Shield",  "Potion",  "Map",  "Torch"]
index:          0          1          2         3        4
negative:       -5         -4         -3        -2       -1
```

```python
print(inventory[0])   # "Sword"   — first item
print(inventory[2])   # "Potion"  — third item
print(inventory[-1])  # "Torch"   — last item (negative index!)
print(inventory[-2])  # "Map"     — second-to-last
```

⚠️ Accessing an index that doesn't exist causes an **IndexError**:
```python
print(inventory[10])  # ❌ IndexError: list index out of range
```

---

## 📖 Concept: Slicing

**Slicing** lets you grab a portion of a list:

```python
list[start : end : step]
```
- `start` — index to begin (inclusive)
- `end`   — index to stop (exclusive, NOT included)
- `step`  — how many to skip (optional)

```python
inventory = ["Sword", "Shield", "Potion", "Map", "Torch"]

print(inventory[1:4])   # ['Shield', 'Potion', 'Map']
print(inventory[:3])    # ['Sword', 'Shield', 'Potion']  — from start
print(inventory[2:])    # ['Potion', 'Map', 'Torch']     — to end
print(inventory[::-1])  # ['Torch', 'Map', 'Potion', 'Shield', 'Sword']  — reversed!
```

````python
# 🐉 Demo: Dungeon loot chest
loot = ["Gold Coin", "Ruby", "Dragon Scale", "Elixir", "Rune Stone", "Crystal"]

print("First loot:", loot[0])
print("Third loot:", loot[2])
print("Last loot:", loot[-1])
print("Second to last:", loot[-2])
print("Items 2-4:", loot[1:4])
print("First 3:", loot[:3])
print("Last 3:", loot[-3:])
print("Reversed:", loot[::-1])
````

# ⚔️ Section 3 — Modifying Lists
**⏱ ~20 minutes**

## 📖 Concept: Lists Are Mutable!

Unlike strings, lists can be **changed** after creation. This is called **mutability**.

### Changing an item by index
```python
inventory = ["Sword", "Shield", "Potion"]
inventory[1] = "Dragon Shield"  # Replace Shield with Dragon Shield
print(inventory)  # ['Sword', 'Dragon Shield', 'Potion']
```

### Adding items
```python
inventory.append("Elixir")      # Add to the END
inventory.insert(1, "Dagger")   # Insert at index 1
```

### Removing items
```python
inventory.remove("Potion")  # Remove by VALUE (first occurrence)
inventory.pop()             # Remove and return the LAST item
inventory.pop(0)            # Remove and return item at index 0
```

| Method | What it does |
|--------|-------------|
| `append(x)` | Adds `x` to the end |
| `insert(i, x)` | Inserts `x` at index `i` |
| `remove(x)` | Removes first occurrence of `x` |
| `pop()` | Removes & returns the last item |
| `pop(i)` | Removes & returns item at index `i` |

````python
# 🐉 Demo: Managing your knight's inventory
inventory = ["Sword", "Shield", "Potion"]
print("Start:", inventory)

# Upgrade Shield
inventory[1] = "Dragon Shield"
print("After upgrade:", inventory)

# Found a new item!
inventory.append("Elixir")
print("After finding Elixir:", inventory)

# Insert a Dagger at the front
inventory.insert(0, "Dagger")
print("After inserting Dagger:", inventory)

# Used a potion (remove it)
inventory.remove("Potion")
print("After using Potion:", inventory)

# Drop the last item
dropped = inventory.pop()
print(f"Dropped: {dropped}")
print("Final inventory:", inventory)
````

# 🏃 Section 4 — Looping Through Lists
**⏱ ~15 minutes**

## 📖 Concept: Iterating Over Lists

You can use a `for` loop (which you already know!) to go through every item in a list.

### Method 1: Loop over items directly
```python
for item in inventory:
    print(item)
```

### Method 2: Loop using `range(len(...))`
Useful when you need the **index**:
```python
for i in range(len(inventory)):
    print(i, inventory[i])
```

### Method 3: Loop using `enumerate()` ⭐ (Best of both worlds!)
Gets both the **index** and the **item** at the same time:
```python
for i, item in enumerate(inventory):
    print(f"{i}: {item}")
```

### Combining loops with `if-else`
```python
for item in inventory:
    if item == "Potion":
        print("Found a potion!")
    else:
        print(f"{item} is not a potion.")
```

````python
# 🐉 Demo: Check the party's readiness
party = ["Aria", "Blaze", "Cinder", "Drake"]
party_hp = [100, 45, 120, 30]

print("=== Party Status ===")
for i, hero in enumerate(party):
    hp = party_hp[i]
    if hp >= 80:
        status = "✅ Ready"
    elif hp >= 50:
        status = "⚠️ Caution"
    else:
        status = "❌ Critical!"
    print(f"{i+1}. {hero}: {hp} HP — {status}")
````

# 🔮 Section 5 — Useful List Methods
**⏱ ~15 minutes**

## 📖 Concept: Handy List Methods

Python gives us many built-in tools for lists! Here are the most useful ones:

| Method / Function | Description | Example |
|---|---|---|
| `list.sort()` | Sorts list **in place** (ascending) | `heroes.sort()` |
| `list.sort(reverse=True)` | Sorts descending | `scores.sort(reverse=True)` |
| `sorted(list)` | Returns a **new** sorted list | `s = sorted(heroes)` |
| `list.count(x)` | Counts occurrences of `x` | `inv.count("Potion")` |
| `list.index(x)` | Returns index of first `x` | `inv.index("Sword")` |
| `x in list` | Checks if `x` is in list | `"Sword" in inv` |
| `sum(list)` | Sum of all numeric items | `sum([10, 20, 30])` |
| `min(list)` | Smallest item | `min([10, 5, 20])` |
| `max(list)` | Largest item | `max([10, 5, 20])` |

```python
scores = [55, 90, 72, 88, 40, 95]
scores.sort(reverse=True)
print(scores)         # [95, 90, 88, 72, 55, 40]
print(sum(scores))    # 440
print(max(scores))    # 95
print(min(scores))    # 40
print(88 in scores)   # True
```

````python
# 🐉 Demo: Hero Tournament Ranking
hero_scores = [55, 90, 72, 88, 40, 95, 63]
heroes      = ["Aria", "Blaze", "Cinder", "Drake", "Echo", "Frost", "Gale"]

print("Scores (original):", hero_scores)

# Sort scores descending
hero_scores_sorted = sorted(hero_scores, reverse=True)
print("Sorted (best first):", hero_scores_sorted)

print("Top score:", max(hero_scores))
print("Lowest score:", min(hero_scores))
print("Total points:", sum(hero_scores))
print("Is Blaze's 90 in scores?", 90 in hero_scores)
print("Cinder's score index:", hero_scores.index(72))
````

# ✨ Bonus Section — List Comprehension
**⏱ ~10 minutes**

## 📖 Concept: Build Lists in One Line

A **list comprehension** is a short way to create a new list from an existing sequence.

```python
new_list = [expression for item in sequence]
```

This loop:
```python
double_damage = []
for damage in hit_damages:
    double_damage.append(damage * 2)
```

can be written as:
```python
double_damage = [damage * 2 for damage in hit_damages]
```

You can also add a condition to keep only certain items:
```python
powerful_hits = [damage for damage in hit_damages if damage >= 75]
```

💡 Read a list comprehension from left to right: **what to store**, **where values come from**, and optionally **which values to keep**.
